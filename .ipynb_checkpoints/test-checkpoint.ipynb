```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# NumPy
arr = np.array([1, 2, 3, 4])
print(arr)

# Pandas
df = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Marks": [80, 85, 90]
})
print(df)

# Matplotlib
plt.plot([1,2,3], [4,5,6])
plt.show()

# Seaborn
sns.barplot(x="Name", y="Marks", data=df)
plt.show()
```

```python
import numpy as np

# 1D Array
arr = np.array([10, 20, 30, 40, 50])
print("Array:", arr)

# Mean, Max, Min
print("Mean:", arr.mean())
print("Max:", arr.max())
print("Min:", arr.min())

# Reshape
matrix = np.array([1,2,3,4,5,6]).reshape(2,3)
print("Reshaped Matrix:\n", matrix)
```

```python
import pandas as pd

data = {
    "Name": ["Rahim", "Karim", "Salma", "Nusrat"],
    "Department": ["CSE", "EEE", "CSE", "BBA"],
    "CGPA": [3.80, 3.50, 3.90, 3.40]
}

df = pd.DataFrame(data)

# Show data
df
```

```python
import matplotlib.pyplot as plt

names = df["Name"]
cgpa = df["CGPA"]

plt.figure()
plt.bar(names, cgpa)
plt.title("CGPA of Students")
plt.xlabel("Name")
plt.ylabel("CGPA")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure()
sns.barplot(x="Name", y="CGPA", data=df)
plt.title("Student CGPA Analysis")
plt.show()
```
